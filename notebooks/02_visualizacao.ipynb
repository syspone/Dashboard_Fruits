Objetivo: criar o gráfico de barras agrupadas e exportá-lo para `reports/figures/`.

> Toda a lógica fica no pacote `src/datafruit`. O notebook só **chama** essas funções e mostra os resultados — assim o mesmo código serve para o notebook, para o pipeline e para o dashboard.

````python
from datafruit.data import load_fruits
from datafruit.visualization import fruits_by_city_chart, save_figure

df = load_fruits()
````

## 1. Gráfico interativo
Cada cor representa uma cidade; as barras ficam lado a lado (`barmode='group'`).

````python
fig = fruits_by_city_chart(df)
fig.show()
````

## 2. Exportar como imagem
O PNG é usado no README do projeto.

````python
save_figure(fig)
````
