## 1. Carregar os dados
Os dados originais ficam em `data/raw/fruits.csv` e **nunca** são editados à mão.

```python
from datafruit.data import load_fruits, save_processed, total_by

df = load_fruits()
df
```

## 2. Qualidade dos dados
Tipos das colunas, valores nulos e duplicados.

```python
df.info()
```

```python
print('Nulos por coluna:')
print(df.isna().sum())
print('\nLinhas duplicadas:', df.duplicated().sum())
```

```python
df.describe()
```

## 3. Perguntas da análise

### Qual é a quantidade total de frutas?

```python
df['Amount'].sum()
```

### Qual fruta tem a maior quantidade?

```python
by_fruit = total_by(df, 'Fruits')
by_fruit
```

### Qual cidade tem a maior quantidade?

```python
by_city = total_by(df, 'City')
by_city
```

### Tabela cruzada: fruta × cidade

```python
df.pivot_table(index='Fruits', columns='City', values='Amount', aggfunc='sum', fill_value=0)
```

## 4. Salvar os resumos
Os resultados tratados vão para `data/processed/`.

```python
save_processed(by_fruit, 'total_by_fruit')
save_processed(by_city, 'total_by_city')
```

## Conclusões

- Total geral: **18 unidades**.
- **Melão** é a fruta com maior quantidade (**7 unidades**).
- **Curitiba** lidera com **9 unidades**; **São Luís** tem a menor (**3 unidades**).
- Não há valores nulos nem linhas duplicadas.
